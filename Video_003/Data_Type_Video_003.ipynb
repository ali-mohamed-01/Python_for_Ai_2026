## Datatypes
- Datatype is set of data with predefined values. The Type of data stored in a variable.

### Built-in Data Types
- In programming, data type is an important concept. Variables can store data of different types, and different types can do different things. Python has the following data types built-in by default, in these categories:

| | |
| :- | :- |
| Text Type | str |
| Numeric Types | int, float |
| Sequence Type | list, tuple, range |
| Mapping Type | Dictionary |
| Set Type | set |
| Boolean Type | bool |
| Binary Types | bytes, bytearray, memoryview |
None Type | NoneType |

### Get the Type
You can get the data type of a variable with the type() function.


```python
x = "Start Now"
print(x.upper())
print(x.lower())

```

```python
y = "Storing text, names, and sentences"
print(y.split())
print(y.split("names"))
```

```python
# ==========================================================
# FULL GUIDE: PYTHON DATA TYPES (BEGINNER TO PRO)
# ==========================================================

# --- 1. Text Type (str) ---
# Store text, names, or sentences
greeting = "Welcome Back"
print(greeting)
print(type(greeting))
```

```python

# --- 2. Numeric Types (int, float) ---
# Floats follow the IEEE 754 standard for floating-point math.
my_int = 10
y_int = -10
x_float = 10.5
y_float = -10.5
print(my_int)
print(type(y_int))
print(y_float)
print(type(x_float))
```

```python

# --- 3. Sequence Types (list, tuple, range) ---
# - Lists [] are ordered and changeable.

my_list = ["data", "analyst", 2026]

print(my_list)
print(my_list[0])
print(type(my_list))
```

```python
# ==============================
# insert (add an element at a specific index)
# ==============================

my_list.insert(1, "science")
# inserts "science" at index 1

print("After insert:", my_list)
```

```python
# ==============================
# remove (delete element by value)
# ==============================

my_list.remove("analyst")
# removes the first occurrence of the value

print("After remove:", my_list)
```

```python
# - Tuples () are ordered but UNCHANGEABLE.

my_tuple = ("fixed", "values")

print(my_tuple)
print(my_tuple[1])
print(type(my_tuple))
```

```python
# - Range generates a sequence of numbers.

my_range = range(1, 10)

print(my_range)
print(my_range[2:7])
print(type(my_range))
```

```python

# --- 4. Mapping Type (dict) ---
# Beginner: Dictionaries store data in Key:Value pairs.

my_dict = {"name": "Ali", "age": "30"}
print(my_dict)
print(type(my_dict))
```

```python
# --- 5. Set Type (set) ---
# Beginner: Sets are unordered collections of UNIQUE items.
# mathematical operations like Union or Intersection at high speed.
my_set = {1, 2, 2, 3, 3} # Duplicates are removed automatically
print(my_set) # Output: {1, 2, 3}
print(type(my_set))
```

```python

# --- 6. Boolean Type (bool) ---
# Beginner: Represents logical values: True or False.

is_valid = True
print(is_valid)
print(type(is_valid))
```

```python

# --- 7. Binary Types (bytes, bytearray, memoryview) ---
# Beginner: Used to handle raw binary data (like images or encrypted text).
# Professional:
# - bytes: Immutable sequence of bytes.
# - bytearray: Mutable (changeable) version of bytes.
# - memoryview: Advanced tool to access data without copying it in memory.
my_bytes = b"Hello"
my_bytearray = bytearray(b"Hello")
my_view = memoryview(my_bytes)
print(my_bytes)
print(type(my_bytearray))
```

```python


# --- 8. None Type (NoneType) ---
# Beginner: Represents the None of a value.

empty_val = None
print(empty_val)
print(type(empty_val))
```

```python
# ==============================================================================
# FINAL SUMMARY TABLE: PYTHON DATA TYPES
# ==============================================================================
# Type          | Mutable?  | Best Usage Scenario
# ------------------------------------------------------------------------------
# str           | No        | Storing text, names, and sentences.
# int           | No        | Counting, indexing, and whole number math.
# float         | No        | Measurements and precise decimal calculations.
# list          | Yes       | Storing collections that need updating or sorting.
# tuple         | No        | Storing fixed data (safety & faster performance).
# range         | No        | Generating number sequences for loops (Memory efficient).
# dict          | Yes       | Fast data lookup using unique Keys (Profiles/JSON).
# set           | Yes       | Removing duplicates and mathematical set operations.
# bool          | No        | Logical conditions (True/False) and flow control.
# bytes         | No        | Handling raw binary data (images, file streams).
# bytearray     | Yes       | Modifying raw binary data without re-allocating.
# memoryview    | No        | Accessing large data buffers without copying.
# NoneType      | No        | Representing "null", empty, or default values.
# ==============================================================================
```
