## Variable
- Variables are containers for storing data values.
- A variable is created the moment you first assign a value to it.
- Python has no command for declaring a variable. We don't need to be declared it with any type (datatype). We can even change type after they have been set.


### Assigning Variables

Listed are several rules for variable names in Python.
* Names can not start with a number
* Names can not contain spaces, use _ instead
* Names can not contain any of these symbols  :'",<>/?|\!@#%^&*~-+
* It's considered best practice ([PEP8](https://www.python.org/dev/peps/pep-0008/#function-and-variable-names)) that names are lowercase letters with underscores
* Avoid using Python built-in keywords like `list` and `str`
* Avoid using the single characters `l` (lowercase letter el), `O` (uppercase letter oh) and `I` (uppercase letter eye) as they can be confused with `1` and `0`

Variable assignment follows `name = object`, where a single equals sign `=` is an **assignment operator**.

```python
# ==========================================
# 2. Assigning Variables - Naming Conventions
# ==========================================
# --- Normal (Flat Case / Simple) ---
# All letters are lowercase with no separators.
username = "Ahmed"
totalamount = 100
isactive = True

# Professional use of f-strings for output:
print("Normal Case: " + username + "Amount:"+ totalamount + " Active: " + isactive)
```

```python
# --- Snake Case (The Python Standard - PEP8) ---
# All letters are lowercase and words are separated by underscores.
user_first_name = "Ali"
total_amount_value = 100
is_logged_in = True

# The Traditional Way (Manual Conversion)
print("Snake Case: " + user_first_name + " Amount: " + str(total_amount_value) + " Status: " + str(is_logged_in))
```

```python

# --- Camel Case ---
# The first word is lowercase, and each subsequent word starts with a capital letter.
userFirstName = "Omar"
totalAmountValue = 100
isPlay = True

# Professional use of f-strings for output:
print(f"Camel Case: {userFirstName} Amount: {totalAmountValue} Playing: {isPlay}")
```
