```python
# ============================================================
#         Import Library
# ============================================================

import operator
import math
from functools import reduce

print("=" * 40)
print("     Arithmetic Operations in Python")
print("=" * 40)
```

```python


# ============================================================
# 1. Addition(Sum)
# ============================================================
print("\n--- Addition ---")

# Level 1:  direct operation
print("Level 1: ", 2 + 1)
```

```python

# Level 2: using variables
a, b = 5, -8
result = a + b
print("Level 2: ", result)
```

```python

# Level 3: using built-in function
print("Level 3: ", sum(25, -11))
```

```python

# Level 4: using operator module
print("Level 4: ", operator.add(-12, -23))
```

```python

# Level 5: using lambda
add = lambda x, y: x + y
print("Level 5: ", add(2, 1))
```

```python
# ============================================================
# 2. Subtraction
# ============================================================
print("\n--- Subtraction ---")

# Level 1: beginner
print("Level 1: ", 9 - 3)
```

```python

# Level 2: using variables
a, b = 5, 1
print("Level 2: ", a - b)
```

```python

# Level 3: using operator
print("Level 3: ", operator.sub(-35, 10))
```

```python

# Level 4: using lambda
subtract = lambda x, y: x - y
print("Level 4: ", subtract(-3, -18))
```

```python
# ============================================================
# 3. Multiplication
# ============================================================
print("\n--- Multiplication ---")

# Level 1: beginner
print("Level 1: ", 2 * 2)
```

```python

# Level 2: using variables
a, b = 12, 13
print("Level 2: ", a * b)
```

```python

# Level 3: using operator
print("Level 3: ", operator.mul(25, 20))
```

```python

# Level 4: using math.prod (Python 3.8+)
print("Level 4: ", math.prod([5, 8]))
```

```python

# Level 5: using lambda
multiply = lambda x, y: x * y
print("Level 5: ", multiply(3, 3))
```

```python
# ============================================================
# 4. Division
# ============================================================
print("\n--- Division ---")

# Level 1: beginner
print("Level 1: ", 3 / 2)
```

```python

# Level 2: using variables
a, b = 36, 5
print("Level 2: ", a / b)
```

```python

# Level 3: using operator
print("Level 3: ", operator.truediv(9, 2))
```

```python

# Level 4: using lambda
divide = lambda x, y: x / y
print("Level 4: ", divide(15, 3))
```

```python

# ============================================================
# 5. Floor Division
# ============================================================
print("\n--- Floor Division ---")

# Level 1: beginner
print("Level 1:", 18 // 4)
```

```python

# Level 2: using variables
a, b = 5, 2
print("Level 2: ", a // b)
```

```python

# Level 3: using operator
print("Level 3: ", operator.floordiv(-25, 5))
```

```python

# Level 4: using math.floor
print("Level 4: ", math.floor(-48 / -10))
```

```python
# ============================================================
# 6. Modulo
# ============================================================
print("\n--- Modulo ---")

# Level 1: beginner
print(7 % 4)
```

```python

# Level 2: using variables
a, b = 16 , 4
print("Level 1: ", a % b)
```

```python

# Level 3: using operator
print("Level 2: ", operator.mod(28, 3))
```

```python
# ============================================================
# 7. Power
# ============================================================
print("\n--- Power ---")

# Level 1: beginner
print("Level 1: ", 2 ** 3)
```

```python

# Level 2: using variables
base, exp = 2, 5
print("Level 2: ", base ** exp)
```

```python

# Level 3: using pow()
print("Level 3: ", pow(3, 4))
```

```python

# Level 4: using operator
print("Level 4: ", operator.pow(5, 3))
```

```python

# Level 5: modular exponentiation
print("Level 5: ", pow(2, 3, 5))
```

```python
# ============================================================
# 8. Square Root
# ============================================================
print("\n--- Square Root ---")

# Level 1: beginner
print("Level 1: ", 4 ** 0.5)
```

```python

# Level 2: using math.sqrt
print("Level 2:", math.sqrt(8))
```

```python

# Level 3: using pow
print("Level 3: ", pow(16, 0.5))
```

```python

# Level 4: cube root and others
print("Level 4: ", 27 ** (1/3))
print(16 ** (1/4))
```

```python

# Level 5: integer square root
print("Level 5: ", math.isqrt(36))
```

```python
# ============================================================
# 9. Order of Operations
# ============================================================
print("\n--- Order of Operations ---")

# Level 1: beginner (PEMDAS rules)
print("Level 1: ", 2 + 10 * 10 + 3)

```

```python

# Level 2: using parentheses
print("Level 2: ", (2 + 10) * (10 + 3))
```

```python

# Level 3: priority examples
print(2 ** 3 + 1)
print(10 % 3 * 2)
print(10 - 3 - 2)

```

```python
# Level 4: full expressions
expr1 = 2 + 10 * 10 + 3
expr2 = (2 + 10) * (10 + 3)
expr3 = 2 ** 2 + 3 * 4 - 1
print(f"2 + 10 * 10 + 3 = {expr1}")
print(f"(2+10) * (10+3) = {expr2}")
print(f"2**2 + 3*4 - 1 = {expr3}")
```

```python

# Level 5: dynamic evaluation (use carefully)
expression = "2 + 10 * 10 + 3"
print(eval(expression))
```

```python
# ============================================================
# 10. Additional Operations
# ============================================================
print("\n--- Additional Operations ---")

# Absolute value
print(abs(-5))
print(math.fabs(-5))
```

```python

# Rounding
print(round(3.7))
print(round(3.4))
print(round(3.14159, 2))
print(math.ceil(3.2))
print(math.floor(3.9))
print(math.trunc(3.9))
```

```python

# Min and Max
print(max(2, 7, 4, 1))
print(min(2, 7, 4, 1))
```

```python

# divmod
print(divmod(7, 4))
```

```python

# Complex numbers
print(complex(2, 3))
print(complex(2, 3) + complex(1, 2))
```

```python

# Logarithms
print(math.log(8, 2))
print(math.log10(100))
print(math.log(math.e))
```

```python

# Constants
print(math.pi) # π
print(math.e)  # e
print(math.tau)  # 2π
print(math.inf) # ∞
```
